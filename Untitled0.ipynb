```python
%%writefile /content/generate_ashes_3d.py
import os
import glob
import torch
from PIL import Image

os.environ["ATTN_BACKEND"] = "xformers"
os.environ["SPCONV_ALGO"] = "native"
os.environ["PYTORCH_CUDA_ALLOC_CONF"] = "expandable_segments:True"

from trellis.pipelines import TrellisImageTo3DPipeline
from trellis.utils import postprocessing_utils

images = glob.glob("/content/ashes_input.*")
if not images:
    raise FileNotFoundError("Uploaded image not found")

input_path = images[0]
output_path = "/content/ashes_product.glb"

print("Input:", input_path)
print("Loading TRELLIS model—first download may take several minutes...")

pipeline = TrellisImageTo3DPipeline.from_pretrained(
    "microsoft/TRELLIS-image-large"
)
pipeline.cuda()

image = Image.open(input_path).convert("RGB")

print("Generating 3D structure and texture...")

outputs = pipeline.run(
    image,
    seed=42,
    formats=["mesh", "gaussian"],
    sparse_structure_sampler_params={
        "steps": 8,
        "cfg_strength": 7.5,
    },
    slat_sampler_params={
        "steps": 8,
        "cfg_strength": 3.0,
    },
)

print("Exporting textured GLB...")

glb = postprocessing_utils.to_glb(
    outputs["gaussian"][0],
    outputs["mesh"][0],
    simplify=0.90,
    texture_size=1024,
)

glb.export(output_path)

size_mb = os.path.getsize(output_path) / (1024 * 1024)
print(f"SUCCESS: {output_path}")
print(f"GLB size: {size_mb:.2f} MB")
```

```python
%%writefile /content/generate_burger_multiview.py
import os
import glob
from PIL import Image

os.environ["ATTN_BACKEND"] = "xformers"
os.environ["SPCONV_ALGO"] = "native"
os.environ["PYTORCH_CUDA_ALLOC_CONF"] = "expandable_segments:True"

from trellis.pipelines import TrellisImageTo3DPipeline
from trellis.utils import postprocessing_utils

paths = sorted(glob.glob("/content/burger_views/view_*.png"))
if len(paths) < 3:
    raise RuntimeError("At least three burger views are required")

images = [Image.open(path).convert("RGB") for path in paths]
output_path = "/content/ashes_burger_multiview_hq.glb"

print(f"Loading TRELLIS for {len(images)} views...")
pipeline = TrellisImageTo3DPipeline.from_pretrained(
    "microsoft/TRELLIS-image-large"
)
pipeline.cuda()

print("Reconstructing burger from multiple views...")
outputs = pipeline.run_multi_image(
    images,
    seed=42,
    formats=["mesh", "gaussian"],
    mode="multidiffusion",
    sparse_structure_sampler_params={
        "steps": 16,
        "cfg_strength": 7.5,
    },
    slat_sampler_params={
        "steps": 16,
        "cfg_strength": 3.0,
    },
)

print("Baking 2048px texture and exporting...")
glb = postprocessing_utils.to_glb(
    outputs["gaussian"][0],
    outputs["mesh"][0],
    simplify=0.70,
    texture_size=2048,
)

glb.export(output_path)
size = os.path.getsize(output_path) / 1024 / 1024

print(f"SUCCESS: {output_path}")
print(f"GLB size: {size:.2f} MB")
```

```python
from google.colab import files
from PIL import Image
from pathlib import Path
import shutil

uploaded = files.upload()

if not uploaded:
    raise RuntimeError("No image was uploaded")

name, data = next(iter(uploaded.items()))
sheet_path = Path("/content/burger_sheet.png")
sheet_path.write_bytes(data)

view_dir = Path("/content/burger_views")
if view_dir.exists():
    shutil.rmtree(view_dir)
view_dir.mkdir(parents=True)

sheet = Image.open(sheet_path).convert("RGB")
w, h = sheet.size
mx, my = w // 2, h // 2
gap = 6

boxes = [
    (0, 0, mx-gap, my-gap),
    (mx+gap, 0, w, my-gap),
    (0, my+gap, mx-gap, h),
    (mx+gap, my+gap, w, h),
]

for index, box in enumerate(boxes, 1):
    view = sheet.crop(box)
    path = view_dir / f"view_{index}.png"
    view.save(path)
    print(path, view.size)

print("CROPPED ALL 4 VIEWS")
```
